# Stable Diffusion v1.5 — `runwayml/stable-diffusion-v1-5`

The full SD 1.5 model (~4 GB). This is the original benchmark model and produces the best quality of the three notebooks.  
**CPU inference is very slow (~10–30 min)**. A GPU is strongly recommended.

| Resource | Requirement |
|---|---|
| Disk space | ~4 GB (in `models/`) |
| RAM (CPU mode) | ~8 GB |
| VRAM (GPU mode) | ~4 GB |

**Important:** launch Jupyter from the project root folder so that `models/` resolves correctly.

```python
import os

cwd = os.getcwd()
if os.path.basename(cwd) == "notebooks":
    project_root = os.path.dirname(cwd)
else:
    project_root = cwd

MODELS_DIR = os.path.join(project_root, "models")
os.makedirs(MODELS_DIR, exist_ok=True)
os.environ["HF_HOME"] = MODELS_DIR

print(f"Models will be cached in: {MODELS_DIR}")
```

```python
from diffusers import StableDiffusionPipeline
import torch

model_id = "runwayml/stable-diffusion-v1-5"

print(f"Loading {model_id} ...")
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float32,
    cache_dir=MODELS_DIR,
)

# --- GPU acceleration ---
# If you have a CUDA GPU, uncomment the next line for 10–20x speedup:
# pipe = pipe.to("cuda")
#
# For Apple Silicon (MPS):
# pipe = pipe.to("mps")

print("Pipeline ready.")
```

## Parameters — feel free to change these

Try the same prompt you used in the tiny and small notebooks to compare output quality across models.

```python
prompt = "a cat wearing a top hat"
num_inference_steps = 25
guidance_scale = 7.5
```

```python
import time

print(f"Generating: '{prompt}'")
print("This may take a while on CPU — grab a coffee.")
t0 = time.time()

image = pipe(
    prompt,
    num_inference_steps=num_inference_steps,
    guidance_scale=guidance_scale,
).images[0]

elapsed = time.time() - t0
print(f"Done in {elapsed:.1f}s")

image
```

## Negative prompts

SD 1.5 supports **negative prompts** — describe what you *don't* want in the image.

```python
prompt = "a cat wearing a top hat, photorealistic, studio lighting"
negative_prompt = "blurry, low quality, cartoon, sketch"

t0 = time.time()
image2 = pipe(
    prompt,
    negative_prompt=negative_prompt,
    num_inference_steps=num_inference_steps,
    guidance_scale=guidance_scale,
).images[0]
print(f"Done in {time.time() - t0:.1f}s")

image2
```

## Save the images (optional)

```python
image.save(os.path.join(project_root, "output_sd15.png"))
image2.save(os.path.join(project_root, "output_sd15_negative.png"))
print("Images saved.")
```
