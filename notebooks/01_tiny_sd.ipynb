# Tiny SD — `segmind/tiny-sd`

A compressed Stable Diffusion model (~500 MB). Runs on CPU in roughly 30–60 seconds.  
Perfect for getting started without a GPU.

**Important:** launch Jupyter from the project root folder so that `models/` resolves correctly.

```python
import os

# Detect project root regardless of where Jupyter was launched from
cwd = os.getcwd()
if os.path.basename(cwd) == "notebooks":
    project_root = os.path.dirname(cwd)
else:
    project_root = cwd

MODELS_DIR = os.path.join(project_root, "models")
os.makedirs(MODELS_DIR, exist_ok=True)
os.environ["HF_HOME"] = MODELS_DIR

print(f"Models will be cached in: {MODELS_DIR}")
```

```python
from diffusers import StableDiffusionPipeline
import torch

model_id = "segmind/tiny-sd"

print(f"Loading {model_id} ...")
pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float32,
    cache_dir=MODELS_DIR,
)

# Uncomment to use GPU if available:
# pipe = pipe.to("cuda")

print("Pipeline ready.")
```

## Parameters — feel free to change these

```python
prompt = "a man and a woman shaking hands"
num_inference_steps = 25   # more steps → better quality, slower
guidance_scale = 5.0       # higher → closer to prompt, less variety
```

```python
import time

print(f"Generating: '{prompt}'")
t0 = time.time()
pipe.safety_checker = None

image = pipe(
    prompt,
    num_inference_steps=num_inference_steps,
    guidance_scale=guidance_scale,
).images[0]

elapsed = time.time() - t0
print(f"Done in {elapsed:.1f}s")

image
```

```python
output_path = os.path.join(project_root, "output_tiny_sd.png")
image.save(output_path)
print(f"Saved to {output_path}")
```
